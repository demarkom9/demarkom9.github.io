```python
from itertools import combinations
import math

points = [(2,3), (5,7), (1,8), (2,3), (9,1), (5,7)]
unique_points=list(set(points))
print(unique_points)

max_dist = 0
result = None

for p1,p2 in combinations(unique_points,2):
    dist = math.dist(p1,p2)
    if dist > max_dist:
     result = (p1,p2,dist)
     max_dist = dist

print(result)
```

```python
class Vehicle:
    def __init__(self, vid, name, speed, battery=100):
        self.id = vid
        self.name = name
        self.speed = speed
        self._battery = max(0, min(100, battery))
        self.status = "idle"

    def move(self, distance):
        if distance <= 0:
            return f"{self.name}: distance must be positive."
        if self._battery <= 10:
            self.status = "idle"
            return f"{self.name} cannot move — low battery!"
        if distance > self._battery:
            self.status = "idle"
            return f"{self.name} doesn't have enough charge."
        self._battery -= distance
        self.status = "moving"
        return f"{self.name} moved {distance} km"

    def charge(self, amount):
        if amount <= 0:
            return f"{self.name}: charge amount must be positive."
        self._battery = min(100, self._battery + amount)
        self.status = "charging"
        return f"{self.name} charged to {self._battery:.1f}%"

    def info(self):
        return f"{self.id} | {self.name} | {self._battery:.1f}% | {self.status}"


v1 = Vehicle(1, "RoboCar", 60)

print(v1.info())
print(v1.move(30))
print(v1.info())
print(v1.move(80))
print(v1.charge(50))
print(v1.info())
```

```python
students = {
    "Alice": [88, 92, 79],
    "Bob": [90, 85, 87],
    "Charlie": [70, 80, 75]
}

def letter_grade(score):
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 70:
        return "C"
    elif score >= 60:
        return "D"
    else:
        return "F"

result = {}

for student_name, grades in students.items():
    average_score = sum(grades) / len(grades)
    result[student_name] = (average_score, letter_grade(average_score))

print(result)
#print(result.items())
sorted_studens = sorted(result.items(),key=lambda x:x[1][0],reverse=True)

print("top student is ", sorted_studens[0])
```

```python
inventory = {
    "laptop": {"price": 1200, "quantity": 4},
    "mouse": {"price": 25, "quantity": 20},
    "keyboard": {"price": 75, "quantity": 10}
}

total_value = 0
value = []

print(inventory.items())

for item, details in inventory.items():
    item_value = details["price"] * details["quantity"]
    value.append((item, item_value))
    total_value += item_value

print("Item values:", value)
print("Total inventory value:", total_value)




```
